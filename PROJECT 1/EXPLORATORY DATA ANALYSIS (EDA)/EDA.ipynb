```python
import pandas as pd

df = pd.read_csv(
    "../DATA CLEANING AND PREPROCESSING/cleaned_adidas_products_for_eda.csv"
)

df.head()
```

```python
#Summary statistics
df[["price", "old_price", "discount", "reviews", "rating"]].describe()
```

```python
df[["price", "old_price", "discount", "reviews", "rating"]].var()
```

```python
#Step 2 — Get the median
df[["price", "old_price", "discount", "reviews", "rating"]].median()
```

```python
#Step 3 — Histograms
import matplotlib.pyplot as plt
import seaborn as sns

df[["price", "discount", "reviews", "rating"]].hist(figsize=(10, 8))

plt.tight_layout()
plt.show()
```

```python

#Step 4 — Scatter Plots
#1. Price vs Discount
plt.figure(figsize=(8, 5))

sns.scatterplot(data=df, x="reviews", y="price")

plt.title("Price vs Reviews")
plt.xlabel("Reviews")
plt.ylabel("Price")
plt.show()
```

```python

#2. Price vs Discount
plt.figure(figsize=(8, 5))

sns.scatterplot(data=df, x="discount", y="price")

plt.title("Price vs Discount")
plt.xlabel("Discount")
plt.ylabel("Price")
plt.show()
```

```python

#3. Rating vs Reviews
plt.figure(figsize=(8, 5))

sns.scatterplot(data=df, x="reviews", y="rating")

plt.title("Rating vs Reviews")
plt.xlabel("Reviews")
plt.ylabel("Rating")
plt.show()
```

```python

#Step 5 — Box Plots
plt.figure(figsize=(10, 6))

sns.boxplot(data=df[["price", "discount", "reviews", "rating"]])

plt.title("Distribution of Numerical Features")
plt.show()
```

```python

#Step 6 — Correlation Matrix
correlation = df[
    ["price", "old_price", "discount", "reviews", "rating"]
].corr()

plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Matrix")
plt.show()
```

```python
#Step 7 — Pair Plots
sns.pairplot(df[["price", "old_price", "discount", "reviews", "rating"]])   
```

## Exploratory Data Analysis Findings

The cleaned Adidas products dataset contains **38 products and 13 variables**.

### 1. Summary Statistics

* The mean product price is **7.02**, while the median is **7.25**.
* The mean discount is **38.65**, with a median of **38.32**.
* The average number of reviews is **11.24**, while the median is **3.5**, indicating that some products have considerably more reviews than others.
* The average rating is **4.65**, with a median of **4.90**, showing generally high product ratings.
* The `old_price` variable has a median of **0** because missing old-price values were represented by zero during preprocessing.

### 2. Distribution

The histograms and box plot show variation across the numerical variables. Reviews have a relatively wide spread, ranging from **0 to 97**, while ratings range from **3.4 to 5.0**.

### 3. Relationship Between Price and Reviews

The scatter plot shows little linear relationship between price and the number of reviews. This is supported by the correlation coefficient of approximately **-0.04**.

### 4. Correlation Analysis

The strongest correlation is between **price and discount (0.90)**, indicating a strong positive linear relationship in this dataset.

There is a moderate positive correlation between **old_price and reviews (0.45)** and between **old_price and discount (0.41)**.

Reviews and ratings have a moderate negative correlation of approximately **-0.37**.

Price and rating have a weak negative correlation of approximately **-0.09**, indicating little linear relationship between the two variables.

### 5. Overall Observation

The EDA reveals substantial variation in product prices, discounts, and review counts. Product ratings are generally high, while the correlation analysis shows that some numerical variables have stronger relationships than others. These findings provide a basis for further analysis of Adidas product characteristics and pricing patterns.
