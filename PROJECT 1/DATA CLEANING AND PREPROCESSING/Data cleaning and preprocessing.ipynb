```python
import requests
from bs4 import BeautifulSoup

url = "https://www.jumia.com.ng/adidas/?q=adidas+sneakers#catalog-listing"

response = requests.get(url)

soup = BeautifulSoup(response.text, "html.parser")


products=soup.find_all("a", class_="core")

Products_Data = []

for product in products:
    name = product.find("h3", class_="name")
    price = product.find("div", class_="prc")
    oldp = product.find("div", class_="old")

    name = name.get_text(strip=True)
    price = price.get_text(strip=True)

    if oldp:
        old_price = oldp.get_text(strip=True)
    else:
        old_price = None

    product_data ={
        "name": name,
        "price": price,
        "old_price": old_price
    }
    

    Products_Data.append(product_data)
print("Products_Data=",Products_Data)
```

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

url = "https://www.jumia.com.ng/mlp-adidas-store/"

response = requests.get(url)

soup = BeautifulSoup(response.text, "html.parser")
products = soup.find_all("a", class_="core")

print("Number of products:", len(products))
```

```python
Products_Data = []

for product in products:

    product_data = {
        "name": product.get("data-ga4-item_name"),
        "brand": product.get("data-ga4-item_brand"),
        "category": product.get("data-ga4-item_category"),
        "category2": product.get("data-ga4-item_category2"),
        "category3": product.get("data-ga4-item_category3"),
        "category4": product.get("data-ga4-item_category4"),
        "product_id": product.get("data-ga4-item_id"),
        "price": product.get("data-ga4-price"),
        "discount": product.get("data-ga4-discount"),
        "reviews": product.get("data-gtm-dimension26"),
        "rating": product.get("data-gtm-dimension27"),
        "position": product.get("data-gtm-position"),
        "old_price": product.find("div", class_="prc").get("data-oprc")
            if product.find("div", class_="prc") else None
    }

    Products_Data.append(product_data)

df = pd.DataFrame(Products_Data)

df.info()
```

```python
df["category4"] = df["category4"].fillna("Unknown")
df["discount"] = df["discount"].fillna("0")
df["old_price"] = df["old_price"].fillna("₦ 0")
```

```python
df = pd.DataFrame(Products_Data)
```

```python
numeric_columns = [
    "price",
    "old_price",
    "discount",
    "reviews",
    "rating",
    "position"
]

for column in numeric_columns:
    df[column] = pd.to_numeric(
        df[column].astype(str).str.replace("₦", "", regex=False).str.replace(",", "", regex=False),
        errors="coerce"
    )
    df.info()
```

```python
df.isnull().sum()
```

```python
df["category4"] = df["category4"].fillna("Unknown")

df["discount"] = df["discount"].fillna(0)

df["reviews"] = df["reviews"].fillna(0)

df["rating"] = df["rating"].fillna(df["rating"].median())

df["old_price"] = df["old_price"].fillna(0)

df.isnull().sum()
```

```python
df[["price", "old_price", "discount", "reviews", "rating"]].describe()
```

```python
Q1 = df["price"].quantile(0.25)
Q3 = df["price"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)
```

```python
outliers = df[(df["price"] < lower_bound) | (df["price"] > upper_bound)]

outliers[["name", "price"]]
```

```python
df_clean = df.copy()
df_clean = df_clean[
    (df_clean["price"] >= lower_bound) &
    (df_clean["price"] <= upper_bound)
]

print("Original rows:", len(df))
print("Rows after outlier removal:", len(df_clean))
```

```python
categorical_columns = [
    "category",
    "category2",
    "category3",
    "category4"
]

df_encoded = pd.get_dummies(
    df_clean,
    columns=categorical_columns,
    dtype=int
)

df_encoded.head()
df_encoded.shape
```
