```python
import requests
from bs4 import BeautifulSoup

url = "https://www.jumia.com.ng/adidas/?q=adidas+sneakers#catalog-listing"

response = requests.get(url)

soup = BeautifulSoup(response.text, "html.parser")

products = soup.find_all("h3", class_="name")
prices = soup.find_all("div", class_="prc")

price_list = []

for price in prices:
    price_list.append(price.get_text(strip=True))

product_names = []

for product in products:
    product_names.append(product.get_text(strip=True))

print("Product_Names = ", product_names)
print("Price_List = ", price_list)
```

```python
products_data = []

for name, price in zip(product_names, price_list):
    product = {
        "name": name,
        "price": price
    }

    products_data.append(product)

print("Products_Data:", products_data)
```

```python
import requests
from bs4 import BeautifulSoup

url = "https://www.jumia.com.ng/adidas/?q=adidas+sneakers#catalog-listing"

response = requests.get(url)

soup = BeautifulSoup(response.text, "html.parser")


products=soup.find_all("a", class_="core")

print( "Products:", products)
print("Number of items found:", len(products))

Products_Data = []

for product in products:
    name = product.find("h3", class_="name")
    price = product.find("div", class_="prc")
    oldp = product.find("div", class_="old")

    name = name.get_text(strip=True)
    price = price.get_text(strip=True)

    if oldp:
        old_price = oldp.get_text(strip=True)
    else:
        old_price = None

    product_data ={
        "name": name,
        "price": price,
        "old_price": old_price
    }
    

    Products_Data.append(product_data)
print("Products_Data=",Products_Data)
print(len(Products_Data))
```

```python
import json

with open("products.json", "w") as file:
    json.dump(Products_Data, file, indent=4)
print("Products_Data:", Products_Data)
```
